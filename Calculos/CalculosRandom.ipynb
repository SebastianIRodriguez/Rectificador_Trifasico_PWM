```kotlin
//Obtener salida requerida del PID de tension para obtener determinada tension en el bus
val Vdc = 350.0
val Rload = 27.07
val P = Vdc.pow(2) / Rload
val Vac = 130 //Vrms
val Iac = P / (3.0 * Vac)
val Id = sqrt(2.0) * Iac
val Vd = sqrt(2.0) * Vac
val voltagePiOutputValue = Id / ((2.0/3.0) * Vdc) * Vd
voltagePiOutputValue

val Req = 0.082 + 0.031
val currentPiOutputValue = Id * Req
currentPiOutputValue
```
